```python
import time
from functools import wraps


def timeit_avg(fn):
    total_sum = 0
    total_cnt = 0

    @wraps(fn)
    def inner(*args, **kwargs):
        nonlocal total_sum, total_cnt

        start = time.time()
        try:
            return fn(*args, **kwargs)
        finally:
            finish = time.time()
            total_sum += finish - start
            total_cnt += 1
            
            print(f"{fn.__name__} avg: {total_sum / total_cnt}")
    return inner


@timeit_avg
def add(a, b):
    return a + b


add(1, 2)
add(1, 5)
add(6, 2)
```

```python
class Num:
    def __init__(self, n):
        self.n = n
        self.__priv = "secret"

    @property
    def priv(self):
        return self.__priv

    def print(self):
        return f"{self.n=}, {self.__priv=}"


class NewNum(Num):
    def print(self):
        return f"{self.n=}, {super().priv=}"


n = NewNum(10)
n.print()
```

```python
class Num:
    def __init__(self, n):
        self.n = n
        self.__priv = "secret"

    def print(self):
        return f"{self.n=}, {self.__priv=}"


class NewNum(Num):
    def __init__(self, n):
        super().__init__(n)
        self.n = n + 99
        self.__priv = "new_secret"

    def print(self):
        return f"{self.n=}, {self.__priv=}"


n = NewNum(10)
n.print()
```
