```python
lst = [1, 2, 3]

def process(obj):
    print(f"{obj=}")
    print(f"111: {sys.getrefcount(obj)=}")
    print(f"222: {sys.getrefcount(lst)=}")

print(f"before: {sys.getrefcount(lst)=}")

process(lst)

print(f"after: {sys.getrefcount(lst)=}")
```

```python
class Student:
    def __init__(self, name, monitor=None):
        self.name = name
        self.monitor = monitor

    def __str__(self):
        return f"Student[name={self.name}, id={id(self)}]"

    def __del__(self):
        print(f"__del__ {str(self)}")
```

```python
steve = Student("steve")
print(steve)

del steve
```

```python
steve = Student("steve")
tom = Student("tom", monitor=steve)
steve.monitor = tom

print(steve, tom)
print(f"{sys.getrefcount(steve)=}, {sys.getrefcount(tom)=}")

del steve, tom
```

```python
import ctypes
```

```python
class PyObject(ctypes.Structure):
    _fields_ = [("refcnt", ctypes.c_long)]


def get_refcnt(obj_id):
    return PyObject.from_address(obj_id).refcnt
```

```python
gc.disable()
```

```python
steve = Student("steve")
tom = Student("tom", monitor=steve)
steve.monitor = tom

id_steve = id(steve)
id_tom = id(tom)

print(steve, tom)
print(f"111: {sys.getrefcount(steve)=}, {sys.getrefcount(tom)=}")
print(f"222: {get_refcnt(id_steve)=}, {get_refcnt(id_tom)=}")

del steve, tom

print(f"333: {get_refcnt(id_steve)=}, {get_refcnt(id_tom)=}")

gc.collect()

print(f"444: {get_refcnt(id_steve)=}, {get_refcnt(id_tom)=}")
```
