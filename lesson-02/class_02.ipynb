```python
from functools import wraps
```

```python
import time


def timeit(fn):
    @wraps(fn)
    def inner(*args, **kwargs):
        start = time.time()
        try:
            return fn(*args, **kwargs)
        finally:
            finish = time.time()
            print(f"{fn.__name__} run: {finish - start}")
    return inner
```

```python
@timeit
def add(a, b):
    time.sleep(1)
    return a + b
```

```python
import time


def timeit_k(k):
    calls = []

    def timeit(fn):
        @wraps(fn)
        def inner(*args, **kwargs):
            start = time.time()
            try:
                return fn(*args, **kwargs)
            finally:
                finish = time.time()
                calls.append(finish - start)
                if len(calls) > k:
                    calls.pop(0)
                print(f"{fn.__name__} {k=} runs: {sum(calls) / len(calls)}")
        return inner
    return timeit
```

```python
@timeit_k(3)
def add(a, b):
    time.sleep(a)
    return a + b
```

```python
for x in range(5):
    add(x, 3)
```
