```python
class AMeta(type):
    def __new__(mcs, name, bases, classdict, **kwargs):
        print(f"AMeta.__new__ {name=}, {bases=}, {classdict=}, {kwargs=}")

        cls = super().__new__(mcs, name, bases, classdict)
        print("AMeta __new__", cls)

        return cls

    def __init__(cls, name, bases, classdict, **kwargs):
        print("AMeta __init__", cls, name, bases, classdict)
        super().__init__(name, bases, classdict, **kwargs)

    def __call__(cls, *args, **kwargs):
        print("AMeta __call__", cls, args, kwargs)
        return super().__call__(*args, **kwargs)

    @classmethod
    def __prepare__(mcs, name, bases, **kwargs):
        print("AMeta __prepare__", mcs, name, bases, kwargs)
        
        return {"added_attr_a": 2, "added_attr_b": 99}
```

```python
class Person(metaclass=AMeta):
    age = 99

    def __new__(cls, *args, **kwargs):
        print("Person __new__", cls, *args, **kwargs)
        return super().__new__(cls)

    def __init__(self, name):
        print("Person __init__", name)
        self.name = name
```

```python
class Singleton:
    _instance = None

    def __new__(cls, *args, **kwargs):
        print("Singleton.__new__", cls)
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


class DbConnector(Singleton):
    def __init__(self, name):
        print("DbConnector.__init__", name)
        self.name = name
```

```python
class MetaSingleton(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        print(f"MetaSingleton.call {cls=}, {args=}, {kwargs=}")

        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)

        return cls._instances[cls]


class DbConnector(metaclass=MetaSingleton):
    def __init__(self, name):
        print("DbConnector.__init__", name)
        self.name = name
```

```python
sqlite = DbConnector("sqlite")
```

```python
mysql = DbConnector("mysql")
```

```python
class CheckMeta(type):
    def __new__(mcs, name, bases, attrs):
        print(f"CheckMeta.__new__ {mcs=}, {name=}, {bases=}, {attrs=}")

        word = "stop"

        for name, val in attrs.items():
            if word in name.lower():
                raise TypeError(f"wrong attr name={name}")

        return super().__new__(mcs, name, bases, attrs)
```

```python
from abc import ABC, ABCMeta, abstractmethod
```

```python
class BasePerson(ABC):
    @abstractmethod
    def eat(self, food):
        print(f"BasePerson.eat {food=}")

    @abstractmethod
    def sleep(self):
        raise NotImplemented

    def study(self):
        self.eat("bread")
        self.sleep()
```

```python
steve = BasePerson()
```

```python
class Person(BasePerson):
    def sleep(self):
        print("Person.sleep")
```

```python
steve = Person()
```

```python
class Student(Person):
    def eat(self, food):
        super().eat(food)
        print("Student.eat", food)
```

```python
steve = Student()
```

```python
steve.eat("apple")
```

```python
steve.study()
```

```python
class Predictable(metaclass=ABCMeta):
    @abstractmethod
    def predict(self):
        pass

    @classmethod
    def __subclasshook__(cls, C):
        #print(f"Predictable.__subclasshook__ {cls=}, {C=}")

        if cls is not Predictable:
            return NotImplemented

        name = "predict"
        for class_ in C.__mro__:
            if name in class_.__dict__ and callable(class_.__dict__[name]):
                return True

        return False


class TrainPredictable(Predictable):
    @abstractmethod
    def train(self):
        pass

    @classmethod
    def __subclasshook__(cls, C):
        if cls is not TrainPredictable:
            return NotImplemented

        for name in ("predict", "train"):
            for class_ in C.__mro__:
                if name in class_.__dict__ and callable(class_.__dict__[name]):
                    break
            else:
                return False

        return True


class RandomForestModel:
    def train(self):
        pass

    def predict(self):
        pass


class TreeModel:
    def predict(self):
        pass


class TrafficLight:
    predict = "green"
    

tree = TreeModel()
traffic = TrafficLight()
forest = RandomForestModel()


print(
    f"{isinstance(tree, Predictable)=}, "
    f"{isinstance(traffic, Predictable)=}, "
    f"{isinstance(forest, Predictable)=}, "
    f"{isinstance(steve, Predictable)=}"
)

print(
    f"{isinstance(tree, TrainPredictable)=}, "
    f"{isinstance(traffic, TrainPredictable)=}, "
    f"{isinstance(forest, TrainPredictable)=}, "
    f"{isinstance(steve, TrainPredictable)=}"
)
```
