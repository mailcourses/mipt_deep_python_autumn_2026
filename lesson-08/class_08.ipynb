```python
class SpecialIterator:
    def __init__(self, limit):
        self.limit = limit
        self.counter = 0

    def __iter__(self):
        print("iter")
        return self
        
    def __next__(self):
        print("next")
        if self.counter < self.limit:
            self.counter += 1
            return self.counter
        else:
            raise StopIteration
```

```python
it = SpecialIterator(5)
```

```python
next(it)
```

```python
def grep(pattern):
   print("start grep for", pattern)
   while True:
       s = yield
       if pattern in s:
           print("found!", s)
       else:
           print("no %s in %s" % (pattern, s))
```

```python
def stepped_counter(start, stop, step):
    print(f"stepped_counter {locals()=}")

    current = start
    while current < stop:
        step_cand = yield current
        if isinstance(step_cand, int):
            step = step_cand

        current += step


counter = stepped_counter(-5, 10, 1)
assert next(counter) == -5
assert next(counter) == -4
assert counter.send(4) == 0
assert next(counter) == 4
assert counter.send(-3) == 1
assert next(counter) == -2
assert counter.send(5) == 3
assert next(counter) == 8
print("ok")

next(counter)
```

```python
import asyncio
import time
```

```python
async def say_after(delay, what):
    await asyncio.sleep(delay)
    print(what)


async def main():
    t1 = time.time()
    await say_after(1, "hello")
    await say_after(2, "world")
    t2 = time.time()
    print("total time", t2 - t1)


await main()
```

```python
async def say_after(delay, what):
    print("start say", what)
    await asyncio.sleep(delay)
    print(what)


async def main():
    t1 = time.time()
    
    task1 = asyncio.create_task(say_after(1, "hello"))
    task2 = asyncio.create_task(say_after(2, "world"))

    await asyncio.gather(task1, task2)
    
    t2 = time.time()
    print("total time", t2 - t1)


await main()
```

```python
async def say_after(delay, what):
    print("start say", what)
    for i in range(5):
        print(f"in range for {what=}, {i=}")
        time.sleep(delay / 5)
        await asyncio.sleep(0)
    print(what)


async def main():
    t1 = time.time()

    task1 = asyncio.create_task(say_after(1, "hello"))
    task2 = asyncio.create_task(say_after(2, "world"))

    await asyncio.gather(task1, task2)
    
    t2 = time.time()
    print("total time", t2 - t1)


await main()
```
