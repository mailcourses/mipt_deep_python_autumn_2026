```python
import sys
import weakref
```

```python
class Student:
    def __init__(self, name, monitor=None):
        self.name = name
        self.monitor = monitor

    def __str__(self):
        return f"Student[name={self.name}, id={id(self)}]"

    def __del__(self):
        print(f"__del__ {str(self)}")
```

```python
steve = Student("steve")
```

```python
steve = Student("steve")
weakref.finalize(steve, lambda: print("final steve"))

print(f"111: {sys.getrefcount(steve)=}")

wr = weakref.ref(steve)
wproxy = weakref.proxy(steve)

print(f"222: {sys.getrefcount(steve)=}")

steve2 = wr()
print(f"333: {sys.getrefcount(steve)=}", steve2 is steve, str(steve2))

print(f"444: {wproxy.name=}")

del steve, steve2

steve3 = wr()
print(f"555: {steve3=}")

print(f"666: {wproxy.name=}")
```

```python
class CustomFilter(logging.Filter):
    def filter(self, rec):
        return rec.args[0] != "steve"
```
